```python
# 1. Loading data

adult_data = pd.read_csv("adult.csv")
```

```python
# 2. Cleaning data

adult_data = adult_data.replace("?", np.nan)
adult_data = adult_data.dropna()
```

```python
# 3. Checking data consistency (salary vs salary K$)

incorrect_rows = adult_data[
    ((adult_data["salary"] == "<=50K") & (adult_data["salary K$"] > 50)) |
    ((adult_data["salary"] == ">50K") & (adult_data["salary K$"] <= 50))
]

if incorrect_rows.shape[0] > 0:
    print(f"Found {incorrect_rows.shape[0]} incorrect rows.")
else:
    print("Data consistency check: PASSED (All rows are correct).")
```

```python
# 4. Demographic Analysis

print("\n--- Gender Distribution ---")
print(adult_data.value_counts("sex"))

mean_male_age = adult_data[adult_data["sex"] == "Male"]["age"].mean()
print(f"\nAverage age of men: {mean_male_age:.1f} years")

total_records = adult_data.shape[0]  # Спростили підрахунок загальної кількості строк
percent_of_polish_residents = (adult_data[adult_data["native-country"] == "Poland"].shape[0] / total_records) * 100
print(f"Percentage of Polish residents: {percent_of_polish_residents:.4f}%")
```

```python
# 5. Education & Salary Analysis

high_education_variants = ["Bachelors", "Prof-school", "Assoc-acdm", "Assoc-voc", "Masters", "Doctorate"]
no_high_education = adult_data[(~adult_data["education"].isin(high_education_variants)) & (adult_data["salary"] == ">50K")]
print(f"\nAmount of people without higher education earning >50K: {no_high_education.shape[0]}")

print("\n--- Age statistics for each education level ---")
print(adult_data.groupby("education", observed=False)["age"].describe())
```

```python
# 6. Marital Status & Income Analysis

married_male = adult_data[adult_data["marital-status"].str.contains("Married") & (adult_data["sex"] == "Male")]
unmarried_male = adult_data[~adult_data["marital-status"].str.contains("Married") & (adult_data["sex"] == "Male")]

print("\n--- Salary distribution among men (%) ---")
print("Married men:")
print(married_male["salary"].value_counts(normalize=True) * 100)
print("\nUnmarried men:")
print(unmarried_male["salary"].value_counts(normalize=True) * 100)
```

```python
# 7. Working Hours Analysis

max_working_hours = adult_data["hours-per-week"].max()
hard_workers = adult_data[adult_data["hours-per-week"] == max_working_hours].shape[0]

print(f"\nMaximum working hours per week: {max_working_hours}")
print(f"Amount of people working max hours: {hard_workers}")
```
